```python
with open(DB_PATH, 'r', encoding = 'utf-8') as f:
    opencookie_db = json.load(f)

print(type(opencookie_db))

print(f"Total entries in database: {len(opencookie_db)}")

keys = list(opencookie_db.keys())
print("\nFirst 5 cookie names")
print(keys[:5])

first_cookiename = keys[0]
print(f"\nData for '{first_cookiename}'--")
print(json.dumps(opencookie_db[first_cookiename], indent=2))
```

```python
flat_cookies = []

for owner, cookie_list in opencookie_db.items():
    for cookie_data in cookie_list:
        cookie_data['owner'] = owner
        flat_cookies.append(cookie_data)

db_df = pd.DataFrame(flat_cookies)

print(f"total individual cookies in database: {len(db_df)}")

print("searching for the 'td' cookie")
print(db_df[db_df['cookie'] == 'td'][['owner', 'cookie', 'domain', 'category']].to_string())
```
